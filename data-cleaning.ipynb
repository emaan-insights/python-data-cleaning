Step 02: Create Employee Dataset

```python
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 105],
    'Name': ['Ali', 'Sara', 'Ahmed', 'Fatima', 'Usman', 'Usman'],
    'Department': ['HR', 'IT', 'hr', 'Finance', None, None],
    'Age': [25, 28, None, 32, 29, 29],
    'Salary': ['50000', '60000', '55000', '70000', '65000', '65000'],
    'Joining_Date': [
        '2023-01-10', '2022-05-15', '2023-03-20',
        '2021-08-01', '2022-11-12', '2022-11-12'
    ]
}

df = pd.DataFrame(data)

df
```

Step 04: Understand the Problem

```python
# Duplicate record         Employee ID 105
# Missing Age              Ahmed 
# Missing Department       Usman
# Inconsistent Text        HR & hr
# Inconsistent datatype    Salary stored as text
# Joining_Date             Date stored as text      
```

Remove Duplicate Row

```python
df = df.drop_duplicates()

df
```

```python
df.duplicated().sum()
```

Handling Missing Values

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
# It fills the missing age using the middle value of the available ages.


df['Department'] = df['Department'].fillna('Unknown')
```

```python
df.isnull().sum()
```

Standardizing Department names

```python
df['Department'] = df['Department'].str.title()

df['Department']
```

```python
df['Department'] = df['Department'].str.upper()

df['Department']
```

Converting Salary into numeric format

```python
df['Salary'] = pd.to_numeric(df['Salary'])

df['Salary'].dtype

#Numeric salary values allow calculations such as average salary and total payroll.
```

Convert Joining date into date format

```python
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'])

df['Joining_Date'].dtype
```

Cleaned Dataset

```python
# Check remaining missing values
df.isnull().sum()
```
